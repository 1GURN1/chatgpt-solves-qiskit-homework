# Instructor Solution - HW3 Seeded Deutsch-Jozsa

This notebook demonstrates the complete reference solution for HW3. It uses `hw3_reference.py` to regenerate the same seeded task and compute the expected output.

```python
%pip -q install qiskit qiskit-aer matplotlib
```

```python
import json
from qiskit.visualization import plot_histogram
from hw3_reference import (
    generate_config, classify_from_mask, expected_input_bits_q0_to_qn,
    expected_display_bitstring, expected_logical_q_string,
    build_deutsch_jozsa_circuit, simulate_reference_counts,
    reference_answers, validate_answers
)
```

```python
STUDENT_ID = "demo_student"
ASSIGNMENT_ID = "HW3"
SHOTS = 2048

config = generate_config(STUDENT_ID, ASSIGNMENT_ID)
print(json.dumps(config, indent=2))
```

## Reference classification and expected outputs

```python
classification = classify_from_mask(config["mask_q0_to_qn"])
expected_bits = expected_input_bits_q0_to_qn(config["mask_q0_to_qn"])
expected_logical = expected_logical_q_string(config)
expected_display = expected_display_bitstring(config)

print("Classification:", classification)
print("Expected input bits q[0]...q[n-1]:", expected_bits)
print("Expected logical q[n-1]...q[0]:", expected_logical)
print("Expected displayed c[n-1]...c[0]:", expected_display)
```

## Build and display reference circuit

```python
qc = build_deutsch_jozsa_circuit(config)
qc.draw("mpl")
```

## Simulator reference counts

```python
counts = simulate_reference_counts(config, shots=SHOTS)
print("Reference counts:", counts)
plot_histogram(counts, title="HW3 Deutsch-Jozsa reference simulator counts")
```

## Reference answers.json fields

```python
ref = reference_answers(STUDENT_ID, shots=SHOTS, assignment_id=ASSIGNMENT_ID)
print(json.dumps(ref, indent=2))
```
